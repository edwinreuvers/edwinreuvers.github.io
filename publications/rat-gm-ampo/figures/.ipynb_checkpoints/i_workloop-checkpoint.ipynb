## Load packages

```python
import os, sys, pickle
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.transforms as mtransforms
from scipy import signal
figDir = os.path.join('../figures','')
funcDir = os.path.join('../functions','')
parDir = os.path.join('../parameters','')
sys.path.append(funcDir)
from CustFig import FigStyle
from FuncMus import ForceEQ, SolveSimuMTC

plt.close('all')
%config InlineBackend.figure_format = 'svg'
```

## Load muscle parameters

```python
mus = 'GMe1'
parFile = os.path.join(parDir,mus+'_CM.pkl')
muspar = pickle.load(open(parFile, 'rb'))[0]
lmtcOpt = (1+muspar['eseerelmax'])*muspar['lsee0']+muspar['lce_opt']
lmtc0 = lmtcOpt+0.5e-3 # [m] MTC-length at t=0
```

## Generate SIN data

```python
cf = 3
amp = 2*1e-3
time = np.linspace(0,5/cf,1000)
lmtc = np.cos(time*2*np.pi*cf)*amp+lmtc0

# Simlation to obtain fsee(t)
gamma0 = muspar['gamma_0']
lcerel0 = ForceEQ(lmtc[0],gamma0,muspar)[1]
c_in = {}
c_in['time'] = time
c_in['lmtc'] = lmtc
c_in['tStim'] = np.array([[0, 0.3/cf], [1/cf, 1.3/cf], [2/cf, 2.3/cf], [3/cf, 3.3/cf], [4/cf, 4.3/cf]])

solstr = SolveSimuMTC(gamma0,lcerel0,muspar,c_in)[1]
fsee = solstr[9]

# Slice data (only last cycle)
iStart = np.argmin(abs(time-4/cf))
iStop = np.argmin(abs(time-5/cf))
time = time[iStart:iStop]-time[iStart]
lmtc = lmtc[iStart:iStop]
fsee = fsee[iStart:iStop]+2
iPeak = signal.find_peaks(-lmtc,distance=200)[0][0]
```

## Make figure

```python
#| label: fig-i-workloop
#| fig-cap: "This is the figure caption."

customlay = FigStyle(fontname='libertinus',fontsize=11)
plt.rcParams.update(plt.rcParamsDefault)
plt.rcParams.update(customlay)
plt.ioff()

fig = plt.figure(figsize=(16/2.54+0.085, 16/3/2.54/1.5+0.085), constrained_layout=True) # 3:1 ratio
gs = fig.add_gridspec(1,3)
axs = [fig.add_subplot(gs[i]) for i in range(0,gs.ncols*gs.nrows)]

# W+
lmtcPos = np.hstack((lmtc[0], lmtc[:iPeak], lmtc[iPeak]))
fseePos = np.hstack((0, fsee[:iPeak], 0))
axs[0].plot(lmtc[:iPeak],fsee[:iPeak],'k')
axs[0].fill(lmtcPos,fseePos,'#bfbfbf')

# W-
lmtcNeg = np.hstack((lmtc[iPeak], lmtc[iPeak:], lmtc[-1]))
fseeNeg = np.hstack((0, fsee[iPeak:], 0))
axs[1].plot(lmtc[iPeak:],fsee[iPeak:],'k')
axs[1].fill(lmtcNeg,fseeNeg,'#bfbfbf')

# Wnet
axs[2].plot(lmtc,fsee,'k')
axs[2].fill(lmtc,fsee,'#bfbfbf')

# Text
axs[0].text(np.median(lmtc),4,'Positive \n work',ha='center',va='center',)
axs[1].text(np.median(lmtc),1,'Negative work',ha='center',va='center',)
axs[2].text(np.median(lmtc),5.3,'Net \n work',ha='center',va='center',)

# Labels
axs[0].set_xlabel('Muscle length')
axs[1].set_xlabel('Muscle length')
axs[2].set_xlabel('Muscle length')
axs[0].set_ylabel('Muscle force')
# axs[1].set_ylabel('Muscle force')
# axs[2].set_ylabel('Muscle force')

# Limits & ticks
axs[0].set_xticks([])
axs[1].set_xticks([])
axs[2].set_xticks([])

axs[0].set_yticks([0])
axs[1].set_yticks([0])
axs[2].set_yticks([0])

axs[0].set_xlim(lmtc.min()-0.0003,lmtc.max()+0.0003)
axs[1].set_xlim(lmtc.min()-0.0003,lmtc.max()+0.0003)
axs[2].set_xlim(lmtc.min()-0.0003,lmtc.max()+0.0003)

axs[0].set_ylim(0,10)
axs[1].set_ylim(0,10)
axs[2].set_ylim(0,10)

# Add the '+' sign between Plot 1 and Plot 2
# axs[0].text(1.1, 0.5, '+', fontsize=30, va='center', ha='center', transform=axs[0].transAxes, weight='bold')

# Add the '=' sign between Plot 2 and Plot 3
# axs[1].text(1.1, 0.5, '=', fontsize=30, va='center', ha='center', transform=axs[1].transAxes)

# Axis labels
labels = [r'\textbf{A}', r'\textbf{B}', r'\textbf{C}']
for (ax,label) in zip(axs, labels):
    # label physical distance to the left and up:
    trans = mtransforms.ScaledTranslation(-12/72, 6/72, fig.dpi_scale_trans)
    ax.text(0.0, 1.0, label, transform=ax.transAxes + trans,
            fontsize='medium', va='top',ha='left',fontweight="bold")

fig.savefig(figDir+'i_workloop.png', dpi=600, bbox_inches="tight", pad_inches=0)
fig.savefig(figDir+'i_workloop.pdf', dpi=600, bbox_inches="tight", pad_inches=0)
```
