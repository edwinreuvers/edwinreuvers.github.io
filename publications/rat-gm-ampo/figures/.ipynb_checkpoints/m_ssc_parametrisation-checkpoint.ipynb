## Load packages

```python
#%% Load packages
import os, sys
import numpy as np
import matplotlib.pyplot as plt
funcDir = os.path.join('../functions')
parDir = os.path.join('../parameters')
sys.path.append(funcDir)
from CustFig import FigStyle
from FuncMotionData import makeSCVdata

plt.close('all')
%config InlineBackend.figure_format = 'svg'
```

## Generate lmtc(t)

```python
cf = 1
fts = 0.65
amp = 1 # [m]
lmtcAvg = 0
acc = 100

time = np.linspace(0,1/cf,1000)
lmtc = makeSCVdata(time,cf,fts,amp,lmtcAvg,acc)[0]
```

## Make figure

```python
#| label: fig-m-ssc-parametrisation
#| fig-cap: "A representation of the parametrisation of the stretch-shortening cycles. $T_{short}$ denotes the *MTC* shortening time, *AMP* denotes the *MTC* amplitude, $L_{avg}^{MTC}$ denotes the average *MTC* length, $T_{leng}$ denotes the *MTC* lengthening time, *CF* denotes the cycle frequency and *FTS* denotes the fraction of the cycle spent shortening (i.e., *MTC* is shortening 65% of the cycle time for a *FTS* of 0.65, as is depicted in this figure)."

customlay = FigStyle(fontname='libertinus',fontsize=11,grid='on')
plt.rcParams.update(customlay)
    
fig = plt.figure(figsize=(8/2.54, 8/3*2/2.54), constrained_layout=True) # 3:2 ratio (1889.763779528 x 1259.8425196853334) px @ 600dpi
fig.set_constrained_layout_pads(w_pad=0, h_pad=0, hspace=0, wspace=0)
gs = fig.add_gridspec(1,1)
axs = [fig.add_subplot(gs[i]) for i in range(0,gs.ncols*gs.nrows)]

ax = axs[0]
ax.plot(time,lmtc,'k')
ax.set_xlim(0,1/cf+0.05)
ax.set_ylim(-1.2,1.1)

axs[0].annotate("", xy=(fts/cf, -1.1), xytext=(0, -1.1),
            arrowprops=dict(arrowstyle="<->"))
text = axs[0].text(fts/cf/2,-1.35, '$T_{Short}$', color='k',ha='center')

axs[0].annotate("", xy=(1, -1.1), xytext=(fts/cf, -1.1),
            arrowprops=dict(arrowstyle="<->"))
text = axs[0].text(fts/cf+(1-fts)/cf/2,-1.35, "$T_{leng}$", color='k',ha='center')

text = axs[0].text(0,-1.8, "$CF = \\frac{1}{T_{short}+T_{leng}}$", color='k',ha='right', va='center') # else part of xlabel is cropped..
text = axs[0].text(1/cf,-1.8, "$FTS = \\frac{T_{short}}{T_{short}+T_{leng}}$", color='k',ha='right', va='center') # else part of xlabel is cropped..

ax.set_xticks([0,fts/cf,1/cf])
ax.set_xticklabels(['','','']) 
ax.set_yticks([-1,0,1])
ax.set_yticklabels(['','$L_{MTC}^{avg}$',''])
ax.set_yticklabels(['$L_{MTC}^{avg} - AMP$','$L_{MTC}^{avg}$','$L_{MTC}^{avg} + AMP$']) 
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

# Save
fig.savefig('m_ssc_parametrisation.png', dpi=600, bbox_inches="tight", pad_inches=0)
fig.savefig('m_ssc_parametrisation.pdf', dpi=600, bbox_inches="tight", pad_inches=0)
```
